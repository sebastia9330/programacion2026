```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough
```

```python
modelo1 = ChatOpenAI(model_name='gpt-3.5-turbo', temperature=0.7, max_tokens=1024)
plantilla = """
    Escriba los puntos mas importantes para {tema}
"""
```

```python
prompt = PromptTemplate(
    input_variables = ['tema'],
    template = plantilla
)

cadena1 = prompt | modelo1
```

```python
modelo2 = ChatOpenAI(model_name='gpt-3.5-turbo', temperature=1.2, max_tokens=1024)
plantilla2 = """
    Necesito que seas un experto guionista y creador de contenido para redes sociales como youtube, necesito 
    que me hagas el script completo para youtube sobre este tema, necesito explicar todo con lujo de detalle:
    {puntos}
"""
```

```python
prompt2 = PromptTemplate(
    input_variables = ['puntos'],
    template = plantilla2
)

cadena2 = prompt2 | modelo2
```

```python
modelo3 = ChatOpenAI(model_name='gpt-4', temperature=1.2, max_tokens=1024)
plantilla3 = '''
    Necesito que seas un experto en SEO, necesito que bajo todo el conocimiento de Youtube que tienes me
    hagas un titulo (100 caracteres incluyendo espacios) y una descripcion (entre 4000 a 5000 caracteres)
    , no anotes los numeros de caracteres, para el script anterior. Que el titulo este en forma de pregunta
    con base a como por ejemplo: ¿cómo vender mas? {script}
'''
```

```python
prompt3 = PromptTemplate(
    input_variables = ['script'],
    template = plantilla3
)

cadena3 = prompt3 | modelo3
```

```python
chainFin = (RunnablePassthrough()
    .assign(puntos = cadena1)
    .assign(script = cadena2)
    .assign(resultadoFinal = cadena3)
)
```

```python
resultado = chainFin.invoke({"tema": 'Ingenieria de Datos'})

print("PUNTOS:\n", resultado["puntos"].content)
print("\nSCRIPT:\n", resultado["script"].content)
print("\nSEO FINAL:\n", resultado["resultadoFinal"].content)
```

#Separar textos

```python
import os
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv(), override = True)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter as RC
```

```python
with open('texto.txt', encoding='utf-8') as r:
    texto = r.read()
```

```python
print(texto)
```

```python
textSplitter = RC(
    chunk_size = 100,
    chunk_overlap = 20,
    length_function = len
)
```

```python
print(textSplitter)
```

```python
textos = textSplitter.create_documents([texto])
print(len(textos))
```

```python
print(textos[0].page_content)
```

```python
import os
from pinecone import Pinecone, ServerlessSpec
from langchain_pinecone import PineconeVectorStore
```

```python
pc = Pinecone(api_key=os.getenv("PINECONE_API_KEY"))
```

```python
indexes = pc.list_indexes()
print(indexes.names())
```

```python
for name in pc.list_indexes().names():
    print('Deleting:', name)
    pc.delete_index(name=name)
```

#crear un nuevo index

```python
index_name = 'textoindex'
if index_name not in pc.list_indexes().names():
    pc.create_index(
        name=index_name,
        dimension=1536,
        metric="cosine",
        spec=ServerlessSpec(
            cloud="aws",
            region="us-east-1"
        )
    )
    print('Hecho')
else:
    print(f'el indice {index_name} ya existe')
```

```python
vector_store = PineconeVectorStore.from_documents(textos,embeddings,index_name=index_name)
```

```python
resultado = vector_store.similarity_search(pregunta)
print(resultado)
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
```

```python
llm = ChatOpenAI(model = 'gpt-3.5-turbo', temperature=1)
```

```python
retriever = vector_store.as_retriever(search_type='similarity', search_kwargs={'k':3})
```

```python
prompt = ChatPromptTemplate.from_template("""
Responde la pregunta usando SOLO el contexto.

Contexto:
{context}

Pregunta:
{question}
""")
```

```python
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)
```

```python
chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough()
    }
    | prompt
    | llm
)
```

```python
response = chain.invoke("¿Que es el regimen juridico aplicable?")
print(response.content)
```
