```python
#importar modulo calendar para obtener el nombre de los meses
import calendar
```

```python
meses = list(calendar.month_name)
meses.pop(0)
```

```python
print(meses)
```

```python
dias = [i for i in range(1,31)]
print(dias)
```

```python
for mes in meses:
    for dia in dias:
        print(f'Mes: {mes}, dia: {dia}')
        ruta = Path(f'2026/{mes}/{dia}')
        ruta.mkdir(parents=True, exist_ok=True)
```
